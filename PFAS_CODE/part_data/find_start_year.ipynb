```python
df_lr_raw = pd.read_csv(path_data_raw + 'lr_raw.csv')
df_sw_raw = pd.read_csv(path_data_raw + 'sw_raw.csv')
dic_inf = {'po':['posname']}
df_lr_raw = mf.append_inf(df_lr_raw, dic_inf)
df_sw_raw = mf.append_inf(df_sw_raw, dic_inf)
df_lr_raw = df_lr_raw[['posname', 'year']][df_lr_raw['type']==0]
df_sw_raw = df_sw_raw[['posname', 'year']][df_sw_raw['type']==0]
df_all = pd.concat([df_lr_raw, df_sw_raw], axis=0)
df_p_y = df_all.groupby(['posname'], as_index=False).agg({'year': 'min'})
df_p_y.rename(columns={'year': 'start_year'}, inplace=True)
df_p_y.to_csv(path_data_raw + 'df_p_y.csv', index=False)

```
