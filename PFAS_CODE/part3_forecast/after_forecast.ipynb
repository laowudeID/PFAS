The first use of some PFAS did not begin in 2000
However, the model does not capture this information
Therefore, additional processing is required after the prediction is complete
Zero in on these Pfas before they are first used

```python
import sys
sys.path.append('..')
import pandas as pd
import numpy as np
import myfunction as mf
from path_config import *

df_po_treat = pd.read_excel(path_file + inf_file, sheet_name="po_treat")

df_treat = df_po_treat[['posname', 'start_year']][df_po_treat['start_year']>2000]

dict_treat = dict(zip(df_treat['posname'], df_treat['start_year']))
print(dict_treat)
list_treat = list(dict_treat.keys())
```

```python

df_adona = pd.read_csv(path_part3_sw + '/sw_lgbm_output/only_pfas/sw_ADONA.csv')
print(df_adona.columns)
df_year = pd.read_csv(path_part3_sw + '/sw_lgbm_output/only_year/sw_2000.csv')
print(df_year.columns)
```

```python

for posname in list_treat:

    file_path = f"{path_part3_sw}/sw_lgbm_output/only_pfas/sw_{posname}.csv"

    df = pd.read_csv(file_path)

    start_year = dict_treat[posname]

    mask = df['year'] < start_year

    columns_to_zero = [col for col in df.columns if col not in ['lon_grid', 'lat_grid', 'year']]

    df.loc[mask, columns_to_zero] = 0

    df.to_csv(file_path, index=False)
```

```python

for posname in list_treat:

    file_path = f"{path_part3_lr}/lr_lgbm_output/only_pfas/lr_{posname}.csv"

    df = pd.read_csv(file_path)

    start_year = dict_treat[posname]

    mask = df['year'] < start_year

    columns_to_zero = [col for col in df.columns if col not in ['lon_grid', 'lat_grid', 'year']]

    df.loc[mask, columns_to_zero] = 0

    df.to_csv(file_path, index=False)

```
