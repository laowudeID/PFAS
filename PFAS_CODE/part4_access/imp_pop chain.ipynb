```python
import sys
sys.path.append('..')
import pandas as pd
import numpy as np
import myfunction as mf
import myaccess as ma
import os
from path_config import *
```

```python

per_wiw_df = pd.read_csv(os.path.join(path_part4_grid, 'per_wiw_grid_year.csv'))
pop_df = pd.read_csv(os.path.join(path_part4_grid, 'population_year.csv'))


pop_data = pop_df.merge(per_wiw_df, on=['lon', 'lat', 'year'], how='left')
pop_data['per_wiw'] = pop_data['per_wiw'].fillna(0)


pop_wiw_dict_lr = {(row['lon'], row['lat'], row['year']): row['pop'] 
                   for _, row in pop_data.iterrows()}

pop_wiw_dict_sw = {(row['lon'], row['lat'], row['year']): row['pop'] * row['per_wiw'] 
                   for _, row in pop_data.iterrows()}
```

### 三种情况单独计算
